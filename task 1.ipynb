# Data Science Internship - Task 1
## Student Performance Data Analysis

**Objective:**  
Analyze the Student Performance dataset (`student-mat.csv`) using Python.  

We will:  
1. Load and explore the dataset  
2. Clean the data  
3. Perform analysis  
4. Answer given questions  
5. Visualize the results  


```python
# Step 1: Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Step 2: Load Dataset  
We load the Student Performance dataset and display the first few rows to understand its structure.


```python
# Load dataset
df = pd.read_csv(r"C:\Users\Mahendra\Downloads\student\student-mat.csv", sep=';')
df.head()
```

## Step 3: Data Exploration  
We now explore the dataset by:  
- Checking missing values  
- Checking data types  
- Viewing dataset shape (rows × columns)  


```python
# Check missing values, data types, and shape
missing_values = df.isnull().sum()
data_types = df.dtypes
shape = df.shape

print("Missing values per column:\n", missing_values)
print("\nData types:\n", data_types)
print(f"\nDataset shape: {shape[0]} rows, {shape[1]} columns")
```

## Step 4: Data Cleaning  
- Handle missing values by filling with the median.  
- Remove duplicate entries.  
This ensures our dataset is consistent and reliable.  


```python
# Fill missing values and remove duplicates
df.fillna(df.median(numeric_only=True), inplace=True)
df.drop_duplicates(inplace=True)

print("Data cleaned successfully!")
```

## Step 5: Data Analysis  
We now answer the analysis questions:  
1. What is the average score in math (G3)?  
2. How many students scored above 15 in their final grade (G3)?  
3. Is there a correlation between study time and the final grade (G3)?  
4. Which gender has a higher average final grade (G3)?  


```python
# Analysis calculations
average_g3 = df['G3'].mean()
count_above_15 = df.loc[df['G3'] > 15, 'G3'].count()
correlation = df['studytime'].corr(df['G3'])
average_g3_gender = df.groupby('sex')['G3'].mean()
higher_avg_gender = average_g3_gender.idxmax()

print(f"Average final grade (G3): {average_g3:.2f}")
print(f"Students scoring above 15: {count_above_15}")
print(f"Correlation between study time and final grade: {correlation:.3f}")
print("\nAverage final grade by gender:\n", average_g3_gender)
print(f"Gender with higher average grade: {higher_avg_gender}")
```

## Step 6: Data Visualization  
We now create the following visualizations:  
1. Histogram of final grades (G3)  
2. Scatter plot between study time and G3  
3. Bar chart comparing male vs female average scores  


```python
# Histogram of final grades (G3)
plt.figure(figsize=(6,4))
plt.hist(df['G3'], bins=10, color='skyblue', edgecolor='black')
plt.title('Distribution of Final Grades (G3)')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')
plt.show()

# Scatter plot between studytime and G3
plt.figure(figsize=(6,4))
sns.scatterplot(x='studytime', y='G3', data=df)
plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time (weekly)')
plt.ylabel('Final Grade (G3)')
plt.show()

# Bar chart comparing average scores by gender
plt.figure(figsize=(6,4))
average_g3_gender.plot(kind='bar', color=['blue','pink'])
plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Final Grade (G3)')
plt.show()
```

# 📌 Summary of Findings
- The average final grade (G3) is around the dataset mean.  
- Several students scored above 15 in their final grade.  
- There is a correlation (positive/negative value shows strength) between study time and grades.  
- Gender comparison shows which gender performs better on average.  

**Visual Insights:**  
- The histogram shows most students score in the middle range.  
- The scatter plot shows the effect of study time on final grades.  
- The bar chart compares male vs female average grades.
